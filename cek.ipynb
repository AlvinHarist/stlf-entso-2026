```python
import numpy as np
import holidays

# Create a copy so we preserve the original plotting dataframe
df_features = df_plot.copy()

# =======================================================
# 1. Time-based Features
# =======================================================
df_features['Hour'] = df_features.index.hour
df_features['DayOfWeek'] = df_features.index.dayofweek  # 0=Mon, 6=Sun
df_features['Month'] = df_features.index.month
df_features['Is_Weekend'] = df_features['DayOfWeek'].apply(lambda x: 1 if x >= 5 else 0)

# Cyclical Encoding for Time (crucial for ML to know Hour 23 is next to Hour 0)
df_features['Hour_sin'] = np.sin(2 * np.pi * df_features['Hour'] / 24.0)
df_features['Hour_cos'] = np.cos(2 * np.pi * df_features['Hour'] / 24.0)
df_features['Month_sin'] = np.sin(2 * np.pi * df_features['Month'] / 12.0)
df_features['Month_cos'] = np.cos(2 * np.pi * df_features['Month'] / 12.0)

# =======================================================
# 2. Holiday Flags (Specifically for Austria)
# =======================================================
at_holidays = holidays.country_holidays('AT')
df_features['Is_Holiday'] = df_features.index.to_series().apply(lambda d: 1 if d.date() in at_holidays else 0)

# =======================================================
# 3. Lag Features
# =======================================================
target = 'AT_load_actual_entsoe_transparency'

# What was the load 1 hour ago?
df_features['Load_Lag_1h'] = df_features[target].shift(1)
# What was the load 24 hours ago (same time yesterday)?
df_features['Load_Lag_24h'] = df_features[target].shift(24)
# What was the load 168 hours ago (same time last week)?
df_features['Load_Lag_168h'] = df_features[target].shift(168)

# =======================================================
# 4. Rolling Averages
# =======================================================
# 24-hour rolling mean of the load (shifted by 1 to prevent predicting the present with the present!)
df_features['Load_Rolling_Mean_24h'] = df_features[target].shift(1).rolling(window=24).mean()
# 24-hour rolling mean for temperature
df_features['Temp_Rolling_Mean_24h'] = df_features['temperature_2m (°C)'].shift(1).rolling(window=24).mean()

# =======================================================
# 5. Drop Missing Values
# =======================================================
# Because we created a 168-hour lag, the first 7 days (168 rows) will have NaN values. 
# We generally drop these initially to have a clean dataset for ML. 
df_features_clean = df_features.dropna()

print("Feature Engineering Complete!")
print(f"Original shape: {df_features.shape}")
print(f"Cleaned shape: {df_features_clean.shape}")

# Review the new features
pd.set_option('display.max_columns', None)
df_features_clean.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

# 1. Define Features (X) and Target (y)
target_col = 'AT_load_actual_entsoe_transparency'

# Drop raw time columns as we are using their sine/cosine and weekend/holiday equivalents
columns_to_drop = ['Hour', 'DayOfWeek', 'Month'] 
X = df_features_clean.drop(columns=[target_col] + columns_to_drop)
y = df_features_clean[[target_col]]

# =======================================================
# 2. Chronological Train / Test Split
# =======================================================
# We will use the first 80% of time-sorted rows for training, and the final 20% for testing
train_size = int(len(df_features_clean) * 0.8)

X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

print(f"Training set size: {X_train.shape[0]} rows")
print(f"Testing set size: {X_test.shape[0]} rows")

# =======================================================
# 3. Data Scaling (MinMaxScaler)
# =======================================================
# We use separate scalers for X and y. Keeping y separate is crucial so we can 
# easily "inverse_transform" our predictions back to actual Megawatts (MW) later.
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()

with open('scaler_x_AT.pkl', 'wb') as f:
    pickle.dump(scaler_X, f)

with open('scaler_y_AT.pkl', 'wb') as f:
    pickle.dump(scaler_y, f)

# FIT on training data only, then TRANSFORM both Train and Test
X_train_scaled_arr = scaler_X.fit_transform(X_train)
X_test_scaled_arr = scaler_X.transform(X_test)

np.save('x_train_scaled_arr_AT.npy', X_train_scaled_arr)
np.save('x_test_scaled_arr_AT.npy', X_test_scaled_arr)

y_train_scaled_arr = scaler_y.fit_transform(y_train)
y_test_scaled_arr = scaler_y.transform(y_test)

np.save('y_train_scaled_arr_AT.npy', y_train_scaled_arr)
np.save('y_test_scaled_arr_AT.npy', y_test_scaled_arr)

# Convert the scaled numpy arrays back into Pandas DataFrames for ease of use
X_train_scaled = pd.DataFrame(X_train_scaled_arr, columns=X.columns, index=X_train.index)
X_test_scaled = pd.DataFrame(X_test_scaled_arr, columns=X.columns, index=X_test.index)

y_train_scaled = pd.DataFrame(y_train_scaled_arr, columns=y.columns, index=y_train.index)
y_test_scaled = pd.DataFrame(y_test_scaled_arr, columns=y.columns, index=y_test.index)

print("Data Splitting and Scaling Complete!")
X_train_scaled.head(3)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import matplotlib.pyplot as plt

# =======================================================
# 1. Train the Baseline Model
# =======================================================
baseline_model = LinearRegression()
baseline_model.fit(X_train_scaled, y_train_scaled)

# =======================================================
# 2. Make Predictions on the Test Set
# =======================================================
y_pred_scaled = baseline_model.predict(X_test_scaled)

# =======================================================
# 3. Inverse Transform to Original Values
# =======================================================
# Convert scaled predictions and actuals back to actual Megawatts (MW)
y_test_actual = scaler_y.inverse_transform(y_test_scaled)
y_pred_actual = scaler_y.inverse_transform(y_pred_scaled)

# =======================================================
# 4. Evaluate the Model
# =======================================================
mae = mean_absolute_error(y_test_actual, y_pred_actual)
rmse = np.sqrt(mean_squared_error(y_test_actual, y_pred_actual))
mape = np.mean(np.abs((y_test_actual - y_pred_actual) / y_test_actual)) * 100

print("=== Baseline Model (Linear Regression) ===")
print(f"Mean Absolute Error (MAE): {mae:.2f} MW")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f} MW")
print(f"Mean Absolute Percentage Error (MAPE): {mape:.2f}%")

# =======================================================
# 5. Visualize the Results (First 7 Days of Test Set)
# =======================================================
# Plotting just the first 168 hours (7 days) so the graph isn't too cluttered
hours_to_plot = 168

plt.figure(figsize=(15, 5))
plt.plot(y_test.index[:hours_to_plot], y_test_actual[:hours_to_plot], label='Actual Load', color='blue', linewidth=2)
plt.plot(y_test.index[:hours_to_plot], y_pred_actual[:hours_to_plot], label='Predicted Load (Baseline)', color='orange', linestyle='--', linewidth=2)
plt.title('Baseline Model: Actual vs Predicted Load (First 7 Days of Test Set)')
plt.xlabel('Time (UTC)')
plt.ylabel('Electrical Load')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# =======================================================
# 1. Reshape Data for LSTM [samples, time_steps, features]
# =======================================================
# All Features Data
X_train_lstm = X_train_scaled.values.reshape((X_train_scaled.shape[0], 1, X_train_scaled.shape[1]))
X_test_lstm = X_test_scaled.values.reshape((X_test_scaled.shape[0], 1, X_test_scaled.shape[1]))

# Load-Only Features Data
X_train_lo_lstm = X_train_lo_scaled.reshape((X_train_lo_scaled.shape[0], 1, X_train_lo_scaled.shape[1]))
X_test_lo_lstm = X_test_lo_scaled.reshape((X_test_lo_scaled.shape[0], 1, X_test_lo_scaled.shape[1]))

# =======================================================
# 2. Define LSTM Model Builder
# =======================================================
def build_lstm_model(input_shape):
    model = Sequential()
    # 50 LSTM units (neurons). You can increase this for complex datasets.
    model.add(LSTM(50, activation='relu', input_shape=input_shape))
    model.add(Dropout(0.2)) # Helps prevent overfitting
    model.add(Dense(1))     # Output layer (1 value: predicted load)
    model.compile(optimizer='adam', loss='mse')
    return model

# Early stopping to stop training when validation loss stops improving
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# =======================================================
# 3. Train Model A: LSTM (ALL Features)
# =======================================================
print("Training LSTM (All Features)...")
lstm_all = build_lstm_model((X_train_lstm.shape[1], X_train_lstm.shape[2]))
lstm_all.fit(X_train_lstm, y_train_scaled, epochs=30, batch_size=64, 
             validation_data=(X_test_lstm, y_test_scaled), 
             callbacks=[early_stop], verbose=1)

y_pred_scaled_lstm_all = lstm_all.predict(X_test_lstm)
y_pred_actual_lstm_all = scaler_y.inverse_transform(y_pred_scaled_lstm_all)

# =======================================================
# 4. Train Model B: LSTM (Load Features ONLY)
# =======================================================
print("\nTraining LSTM (Load Features Only)...")
lstm_lo = build_lstm_model((X_train_lo_lstm.shape[1], X_train_lo_lstm.shape[2]))
lstm_lo.fit(X_train_lo_lstm, y_train_lo_scaled, epochs=30, batch_size=64, 
            validation_data=(X_test_lo_lstm, y_test_lo_scaled), 
            callbacks=[early_stop], verbose=1)

y_pred_scaled_lstm_lo = lstm_lo.predict(X_test_lo_lstm)
y_pred_actual_lstm_lo = scaler_y.inverse_transform(y_pred_scaled_lstm_lo)

# =======================================================
# 5. Evaluate and Compare
# =======================================================
def print_metrics(name, actual, pred):
    mae = mean_absolute_error(actual, pred)
    rmse = np.sqrt(mean_squared_error(actual, pred))
    mape = np.mean(np.abs((actual - pred) / actual)) * 100
    print(f"=== {name} ===")
    print(f"MAE: {mae:.2f} MW | RMSE: {rmse:.2f} MW | MAPE: {mape:.2f}%\n")

print("\n" + "="*40)
print_metrics("LSTM 1: ALL Features (Weather + Time + Load)", y_test_actual, y_pred_actual_lstm_all)
print_metrics("LSTM 2: Load History Features ONLY", y_test_actual, y_pred_actual_lstm_lo)

# =======================================================
# 6. Visualize the Results
# =======================================================
hours_to_plot = 168 # 7 days

plt.figure(figsize=(15, 6))
plt.plot(y_test.index[:hours_to_plot], y_test_actual[:hours_to_plot], label='Actual Load', color='black', linewidth=2)
plt.plot(y_test.index[:hours_to_plot], y_pred_actual_lstm_all[:hours_to_plot], label='LSTM (All Features)', color='green', linestyle='--', linewidth=2)
plt.plot(y_test.index[:hours_to_plot], y_pred_actual_lstm_lo[:hours_to_plot], label='LSTM (Load Only)', color='red', linestyle=':', linewidth=2)

plt.title('LSTM Forecast Comparison: First 7 Days of Test Set')
plt.xlabel('Time (UTC)')
plt.ylabel('Electrical Load (MW)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Update the results DataFrame and CSV
results_df['LSTM_All_Features'] = y_pred_actual_lstm_all.flatten()
results_df['LSTM_Load_Only'] = y_pred_actual_lstm_lo.flatten()
results_df.to_csv('results/forecast_results_combined.csv', index=False)
```

```python
from tensorflow.keras.layers import Input, Dense, Dropout, LayerNormalization, MultiHeadAttention, Add, GlobalAveragePooling1D
from tensorflow.keras.models import Model

# =======================================================
# 1. Define Transformer Model Builder
# =======================================================
def build_transformer_model(input_shape, head_size=64, num_heads=4, ff_dim=64, dropout=0.2):
    inputs = Input(shape=input_shape)
    
    # Transformer Block
    # 1. Multi-Head Self Attention
    x = LayerNormalization(epsilon=1e-6)(inputs)
    attention_output = MultiHeadAttention(key_dim=head_size, num_heads=num_heads, dropout=dropout)(x, x)
    x = Add()([attention_output, inputs]) # Residual connection
    
    # 2. Feed-Forward Network
    x2 = LayerNormalization(epsilon=1e-6)(x)
    x2 = Dense(ff_dim, activation="relu")(x2)
    x2 = Dropout(dropout)(x2)
    x2 = Dense(input_shape[-1])(x2) # Output dimension must match input for residual connection
    x = Add()([x2, x]) # Residual connection
    
    # 3. Output Layers
    x = GlobalAveragePooling1D()(x) # Flattens the time_steps dimension
    x = Dense(32, activation="relu")(x)
    x = Dropout(dropout)(x)
    outputs = Dense(1)(x)
    
    model = Model(inputs=inputs, outputs=outputs)
    model.compile(optimizer="adam", loss="mse")
    return model

# We can reuse the 3D data (X_train_lstm / X_train_lo_lstm) we prepared for the LSTM!

# =======================================================
# 2. Train Model G: Transformer (ALL Features)
# =======================================================
print("Training Transformer (All Features)...")
input_shape_all = (X_train_lstm.shape[1], X_train_lstm.shape[2])
transformer_all = build_transformer_model(input_shape_all)

transformer_all.fit(X_train_lstm, y_train_scaled, epochs=30, batch_size=64, 
                    validation_data=(X_test_lstm, y_test_scaled), 
                    callbacks=[early_stop], verbose=1)

y_pred_scaled_trans_all = transformer_all.predict(X_test_lstm)
y_pred_actual_trans_all = scaler_y.inverse_transform(y_pred_scaled_trans_all)

# =======================================================
# 3. Train Model H: Transformer (Load Features ONLY)
# =======================================================
print("\nTraining Transformer (Load Features Only)...")
input_shape_lo = (X_train_lo_lstm.shape[1], X_train_lo_lstm.shape[2])
transformer_lo = build_transformer_model(input_shape_lo)

transformer_lo.fit(X_train_lo_lstm, y_train_lo_scaled, epochs=30, batch_size=64, 
                   validation_data=(X_test_lo_lstm, y_test_lo_scaled), 
                   callbacks=[early_stop], verbose=1)

y_pred_scaled_trans_lo = transformer_lo.predict(X_test_lo_lstm)
y_pred_actual_trans_lo = scaler_y.inverse_transform(y_pred_scaled_trans_lo)

# =======================================================
# 4. Evaluate and Compare
# =======================================================
print("\n" + "="*40)
print_metrics("Transformer 1: ALL Features (Weather + Time + Load)", y_test_actual, y_pred_actual_trans_all)
print_metrics("Transformer 2: Load History Features ONLY", y_test_actual, y_pred_actual_trans_lo)

# =======================================================
# 5. Visualize the Results
# =======================================================
hours_to_plot = 168 # 7 days

plt.figure(figsize=(15, 6))
plt.plot(y_test.index[:hours_to_plot], y_test_actual[:hours_to_plot], label='Actual Load', color='black', linewidth=2)
plt.plot(y_test.index[:hours_to_plot], y_pred_actual_trans_all[:hours_to_plot], label='Transformer (All Features)', color='cyan', linestyle='--', linewidth=2)
plt.plot(y_test.index[:hours_to_plot], y_pred_actual_trans_lo[:hours_to_plot], label='Transformer (Load Only)', color='brown', linestyle=':', linewidth=2)

plt.title('Transformer Forecast Comparison: First 7 Days of Test Set')
plt.xlabel('Time (UTC)')
plt.ylabel('Electrical Load (MW)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# =======================================================
# 6. Update the results DataFrame and CSV
# =======================================================
results_df['Transformer_All_Features'] = y_pred_actual_trans_all.flatten()
results_df['Transformer_Load_Only'] = y_pred_actual_trans_lo.flatten()
results_df.to_csv('results/forecast_results_combined.csv', index=False)

print("Transformer results added to 'results/forecast_results_combined.csv'")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from vmdpy import VMD

# =======================================================
# 1. Define VMD Parameters
# =======================================================
alpha = 2000       # moderate bandwidth constraint
tau = 0.           # noise-tolerance (no strict fidelity enforcement)
K = 10              # number of modes/IMFs (you can tune this, usually 3 to 8)
DC = 0             # no DC part imposed
init = 1           # initialize omegas uniformly
tol = 1e-7         # tolerance of convergence

# Extract the continuous signal we want to decompose (the electrical load)
# We use the raw target from df_features_clean
target_signal = df_features_clean['AT_load_actual_entsoe_transparency'].values

# =======================================================
# 2. Run VMD
# =======================================================
print(f"Running VMD to extract {K} modes (IMFs)... This might take a moment.")
# u: the collection of decomposed modes
# u_hat: spectra of the modes
# omega: estimated mode center-frequencies
u, u_hat, omega = VMD(target_signal, alpha, tau, K, DC, init, tol)

# =======================================================
# 3. Add IMFs as New Features
# =======================================================
# We shift the IMFs by 1 just like we did with lags earlier! 
# We cannot use the current timestamp's IMF to predict the current timestamp's load (Data Leakage).
for i in range(K):
    df_features_clean[f'Load_IMF_{i+1}_Lag1'] = pd.Series(u[i, :], index=df_features_clean.index).shift(1)

# Drop the first row which now has NaN due to the new Lag1 shift
df_vmd_clean = df_features_clean.dropna()

print(f"VMD complete. {K} new lagged IMF features added to the dataset.")

# =======================================================
# 4. Visualize the Decomposed Modes (IMFs)
# =======================================================
# Let's plot the original signal and its IMFs for the first 1000 hours to see the patterns
hours_to_visualize = 1000

plt.figure(figsize=(15, 2 * (K + 1)))

# Original Signal
plt.subplot(K + 1, 1, 1)
plt.plot(target_signal[:hours_to_visualize], color='black', linewidth=1.5)
plt.title("Original Signal: Actual Electrical Load")
plt.ylabel("MW")
plt.grid(True)

# Plot each regular IMF
colors = plt.cm.viridis(np.linspace(0, 1, K))
for i in range(K):
    plt.subplot(K + 1, 1, i + 2)
    plt.plot(u[i, :hours_to_visualize], color=colors[i], linewidth=1.5)
    plt.title(f"Mode {i+1} (IMF {i+1})")
    plt.ylabel("MW")
    plt.grid(True)

plt.tight_layout()
plt.show()

# Display the updated dataframe with the new VMD features
df_vmd_clean.head()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

# =======================================================
# 1. Define Features & Target for VMD Dataset
# =======================================================
target_col = 'AT_load_actual_entsoe_transparency'

# Define "Load Only VMD" features
load_only_vmd_features = [
    'Load_Lag_1h', 'Load_Lag_24h', 'Load_Lag_168h', 'Load_Rolling_Mean_24h'
] + [f'Load_IMF_{i+1}_Lag1' for i in range(K)]  # Add all K modes

# Dataframes derived from df_vmd_clean
columns_to_drop = ['Hour', 'DayOfWeek', 'Month']
X_vmd_all = df_vmd_clean.drop(columns=[target_col] + columns_to_drop)
X_vmd_lo = df_vmd_clean[load_only_vmd_features]
y_vmd = df_vmd_clean[[target_col]]

# =======================================================
# 2. Chronological Train / Test Split
# =======================================================
train_size = int(len(df_vmd_clean) * 0.8)

X_train_vmd_all_raw, X_test_vmd_all_raw = X_vmd_all.iloc[:train_size], X_vmd_all.iloc[train_size:]
X_train_vmd_lo_raw, X_test_vmd_lo_raw = X_vmd_lo.iloc[:train_size], X_vmd_lo.iloc[train_size:]
y_train_vmd_raw, y_test_vmd_raw = y_vmd.iloc[:train_size], y_vmd.iloc[train_size:]

# =======================================================
# 3. Data Scaling (MinMaxScaler)
# =======================================================
scaler_X_vmd_all = MinMaxScaler()
scaler_X_vmd_lo = MinMaxScaler()
scaler_y_vmd = MinMaxScaler()

# Scale All Features
X_train_vmd_all = scaler_X_vmd_all.fit_transform(X_train_vmd_all_raw)
X_test_vmd_all = scaler_X_vmd_all.transform(X_test_vmd_all_raw)

# Scale Load Only Features
X_train_vmd_lo = scaler_X_vmd_lo.fit_transform(X_train_vmd_lo_raw)
X_test_vmd_lo = scaler_X_vmd_lo.transform(X_test_vmd_lo_raw)

# Scale Target
y_train_vmd_scaled = scaler_y_vmd.fit_transform(y_train_vmd_raw)
y_test_vmd_scaled = scaler_y_vmd.transform(y_test_vmd_raw)
y_test_vmd_actual = y_test_vmd_raw.values

# =======================================================
# 4. Reshape Data for 3D Deep Learning Models
# =======================================================
X_train_vmd_all_3D = X_train_vmd_all.reshape((X_train_vmd_all.shape[0], 1, X_train_vmd_all.shape[1]))
X_test_vmd_all_3D = X_test_vmd_all.reshape((X_test_vmd_all.shape[0], 1, X_test_vmd_all.shape[1]))

X_train_vmd_lo_3D = X_train_vmd_lo.reshape((X_train_vmd_lo.shape[0], 1, X_train_vmd_lo.shape[1]))
X_test_vmd_lo_3D = X_test_vmd_lo.reshape((X_test_vmd_lo.shape[0], 1, X_test_vmd_lo.shape[1]))

# Prepare a dictionary to hold all inverse-transformed predictions for our final DataFrame
vmd_predictions = {}
results_vmd_df = pd.DataFrame({'utc_timestamp': y_test_vmd_raw.index, 'Actual_Load': y_test_vmd_actual.flatten()})

# Utility function to predict, inverse transform, evaluate, and save
def evaluate_and_store(model, X_test, model_name, is_3d=False):
    y_pred_scaled = model.predict(X_test)
    y_pred_actual = scaler_y_vmd.inverse_transform(y_pred_scaled).flatten()
    vmd_predictions[model_name] = y_pred_actual
    results_vmd_df[model_name] = y_pred_actual
    
    mae = mean_absolute_error(y_test_vmd_actual, y_pred_actual)
    rmse = np.sqrt(mean_squared_error(y_test_vmd_actual, y_pred_actual))
    mape = np.mean(np.abs((y_test_vmd_actual.flatten() - y_pred_actual) / y_test_vmd_actual.flatten())) * 100
    print(f"{model_name: <30} | MAE: {mae:.2f} MW | RMSE: {rmse:.2f} MW | MAPE: {mape:.2f}%")

print("="*60)
print("TRAINING AND EVALUATING VMD-INCORPORATED MODELS")
print("="*60)

# --------------------------
# A. Linear Regression
# --------------------------
lr_vmd_all = LinearRegression().fit(X_train_vmd_all, y_train_vmd_scaled)
evaluate_and_store(lr_vmd_all, X_test_vmd_all, 'LinReg_VMD_All')

lr_vmd_lo = LinearRegression().fit(X_train_vmd_lo, y_train_vmd_scaled)
evaluate_and_store(lr_vmd_lo, X_test_vmd_lo, 'LinReg_VMD_LoadOnly')

# --------------------------
# B. MLP (Dense)
# --------------------------
print("\nTraining MLP Models...")
mlp_vmd_all = build_mlp_model(X_train_vmd_all.shape[1])
mlp_vmd_all.fit(X_train_vmd_all, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_all, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(mlp_vmd_all, X_test_vmd_all, 'MLP_VMD_All')

mlp_vmd_lo = build_mlp_model(X_train_vmd_lo.shape[1])
mlp_vmd_lo.fit(X_train_vmd_lo, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_lo, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(mlp_vmd_lo, X_test_vmd_lo, 'MLP_VMD_LoadOnly')

# --------------------------
# C. LSTM
# --------------------------
print("\nTraining LSTM Models...")
lstm_vmd_all = build_lstm_model((1, X_train_vmd_all_3D.shape[2]))
lstm_vmd_all.fit(X_train_vmd_all_3D, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_all_3D, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(lstm_vmd_all, X_test_vmd_all_3D, 'LSTM_VMD_All', is_3d=True)

lstm_vmd_lo = build_lstm_model((1, X_train_vmd_lo_3D.shape[2]))
lstm_vmd_lo.fit(X_train_vmd_lo_3D, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_lo_3D, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(lstm_vmd_lo, X_test_vmd_lo_3D, 'LSTM_VMD_LoadOnly', is_3d=True)

# --------------------------
# D. BiLSTM
# --------------------------
print("\nTraining BiLSTM Models...")
bilstm_vmd_all = build_bilstm_model((1, X_train_vmd_all_3D.shape[2]))
bilstm_vmd_all.fit(X_train_vmd_all_3D, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_all_3D, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(bilstm_vmd_all, X_test_vmd_all_3D, 'BiLSTM_VMD_All', is_3d=True)

bilstm_vmd_lo = build_bilstm_model((1, X_train_vmd_lo_3D.shape[2]))
bilstm_vmd_lo.fit(X_train_vmd_lo_3D, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_lo_3D, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(bilstm_vmd_lo, X_test_vmd_lo_3D, 'BiLSTM_VMD_LoadOnly', is_3d=True)

# --------------------------
# E. Transformer
# --------------------------
print("\nTraining Transformer Models...")
trans_vmd_all = build_transformer_model((1, X_train_vmd_all_3D.shape[2]))
trans_vmd_all.fit(X_train_vmd_all_3D, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_all_3D, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(trans_vmd_all, X_test_vmd_all_3D, 'Transformer_VMD_All', is_3d=True)

trans_vmd_lo = build_transformer_model((1, X_train_vmd_lo_3D.shape[2]))
trans_vmd_lo.fit(X_train_vmd_lo_3D, y_train_vmd_scaled, epochs=30, batch_size=64, validation_data=(X_test_vmd_lo_3D, y_test_vmd_scaled), callbacks=[early_stop], verbose=0)
evaluate_and_store(trans_vmd_lo, X_test_vmd_lo_3D, 'Transformer_VMD_LoadOnly', is_3d=True)

# =======================================================
# 5. Save Results to CSV
# =======================================================
csv_path = 'results/forecast_results_vmd_combined.csv'
results_vmd_df.to_csv(csv_path, index=False)
print(f"\nAll VMD benchmark results saved to '{csv_path}'!")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import TimeSeriesSplit
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_percentage_error
import warnings
warnings.filterwarnings('ignore')

# 1. Define the parameters for the experiment
horizons = [6, 12, 24, 48, 72]
n_splits = 3 # Number of Time Series CV folds
tscv = TimeSeriesSplit(n_splits=n_splits)

target_col = 'AT_load_actual_entsoe_transparency'

# Function to build Deep Learning models (reused from your previous cells)
def get_compiled_model(model_name, input_shape):
    if model_name == 'MLP':
        return build_mlp_model(input_shape[1]) # input_shape[1] is features
    elif model_name == 'LSTM':
        return build_lstm_model((1, input_shape[1]))
    elif model_name == 'BiLSTM':
        return build_bilstm_model((1, input_shape[1]))
    elif model_name == 'Transformer':
        return build_transformer_model((1, input_shape[1]))
    return None

# Datasets dictionary to iterate through
datasets = {
    'Standard': df_features_clean,
    'VMD': df_vmd_clean
}

model_names = ['LinReg', 'MLP', 'LSTM', 'BiLSTM', 'Transformer']
results_list = []

print("Starting Multi-Step Time Series Cross-Validation Experiment...")

# 2. Iterate over Horizons (6h, 12h, 24h, 48h, 72h)
for h in horizons:
    print(f"\n{'='*50}\nForecasting Horizon: {h} Hours Ahead\n{'='*50}")
    
    for data_name, df_base in datasets.items():
        # Create the h-step ahead target by shifting the target column backwards by h
        df_h = df_base.copy()
        df_h[f'Target_{h}h'] = df_h[target_col].shift(-h)
        df_h = df_h.dropna() # Drop the last h rows which now have NaN targets
        
        # Define Features
        if data_name == 'Standard':
            cols_to_drop = ['Hour', 'DayOfWeek', 'Month', target_col, f'Target_{h}h']
            X_all = df_h.drop(columns=[c for c in cols_to_drop if c in df_h.columns])
            X_load = df_h[['Load_Lag_1h', 'Load_Lag_24h', 'Load_Lag_168h', 'Load_Rolling_Mean_24h']]
        else: # VMD
            cols_to_drop = ['Hour', 'DayOfWeek', 'Month', target_col, f'Target_{h}h']
            X_all = df_h.drop(columns=[c for c in cols_to_drop if c in df_h.columns])
            vmd_load_feeds = ['Load_Lag_1h', 'Load_Lag_24h', 'Load_Lag_168h', 'Load_Rolling_Mean_24h'] + [f'Load_IMF_{i+1}_Lag1' for i in range(10)]
            X_load = df_h[vmd_load_feeds]
            
        target_y = df_h[[f'Target_{h}h']]
        
        feature_sets = {'All Features': X_all, 'Load Only': X_load}
        
        for feat_name, X_feat in feature_sets.items():
            print(f"\n--- Data: {data_name} | Features: {feat_name} ---")
            
            # Convert to numpy for CV splitting
            X_arr = X_feat.values
            y_arr = target_y.values
            
            # Cross Validation
            fold = 1
            for train_index, test_index in tscv.split(X_arr):
                X_tr, X_val = X_arr[train_index], X_arr[test_index]
                y_tr, y_val = y_arr[train_index], y_arr[test_index]
                
                # Scale Data
                scaler_X = MinMaxScaler().fit(X_tr)
                scaler_y = MinMaxScaler().fit(y_tr)
                
                X_tr_sc = scaler_X.transform(X_tr)
                X_val_sc = scaler_X.transform(X_val)
                y_tr_sc = scaler_y.transform(y_tr)
                
                # Reshape for 3D models explicitly
                X_tr_3d = X_tr_sc.reshape((X_tr_sc.shape[0], 1, X_tr_sc.shape[1]))
                X_val_3d = X_val_sc.reshape((X_val_sc.shape[0], 1, X_val_sc.shape[1]))
                
                for m_name in model_names:
                    if m_name == 'LinReg':
                        model = LinearRegression().fit(X_tr_sc, y_tr_sc)
                        preds_sc = model.predict(X_val_sc)
                    else:
                        model = get_compiled_model(m_name, X_tr_sc.shape)
                        
                        # FIXED LINE: Explicitly determine 3D requirement based on model name
                        is_3d = m_name in ['LSTM', 'BiLSTM', 'Transformer']
                        
                        feed_tr = X_tr_3d if is_3d else X_tr_sc
                        feed_val = X_val_3d if is_3d else X_val_sc
                        
                        from tensorflow.keras.callbacks import EarlyStopping
                        es = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
                        model.fit(feed_tr, y_tr_sc, epochs=15, batch_size=128, validation_data=(feed_val, scaler_y.transform(y_val)), callbacks=[es], verbose=0)
                        preds_sc = model.predict(feed_val, verbose=0)
                        
                    # Inverse and Evaluate
                    preds_actual = scaler_y.inverse_transform(preds_sc)
                    fold_mape = mean_absolute_percentage_error(y_val, preds_actual) * 100
                    
                    results_list.append({
                        'Horizon': f"{h}h",
                        'Method': data_name,
                        'Feature_Set': feat_name,
                        'Model': m_name,
                        'CV_Fold': fold,
                        'MAPE': fold_mape
                    })
                    print(f"[{m_name}] Fold {fold} MAPE: {fold_mape:.2f}%")
                fold += 1

# 3. Aggregate and Save Results
cv_results_df = pd.DataFrame(results_list)
cv_results_df.to_csv('results/cv_multihorizon_experiment.csv', index=False)

# Average the MAPE across CV folds for the final summary table
summary_cv = cv_results_df.groupby(['Horizon', 'Method', 'Feature_Set', 'Model'])['MAPE'].mean().reset_index()

print("\n=== EXPERIMENT COMPLETE ===")
display(summary_cv.pivot_table(index=['Horizon', 'Model'], columns=['Method', 'Feature_Set'], values='MAPE').round(2))
```
