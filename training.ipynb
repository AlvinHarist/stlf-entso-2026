```python
import os
import numpy as np
import pandas as pd
import datetime
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

# Verify GPU
print("Num GPUs Available: ", len(tf.config.list_physical_devices('GPU')))

# Create Directories
os.makedirs('MODELS', exist_ok=True)
os.makedirs('RESULTS', exist_ok=True)
os.makedirs('LOSS', exist_ok=True)

# Load Dataset
print("Loading data...")
X_train = np.load('dataset_ready/X_train.npy')
y_train = np.load('dataset_ready/y_train.npy')
X_val = np.load('dataset_ready/X_val.npy')
y_val = np.load('dataset_ready/y_val.npy')
X_test = np.load('dataset_ready/X_test.npy')
y_test = np.load('dataset_ready/y_test.npy')

# Assuming date retrieval from df_combined_AT.csv based on standard sliding windows
# Adjust path if needed
try:
    df = pd.read_csv("df_combined_AT.csv")
    total_samples = len(df) - 24 - 24 + 1
    train_end = int(total_samples * 0.8)
    val_end = train_end + int(total_samples * 0.1)
    # Get test start timestamp
    test_timestamps = df['utc_timestamp'].iloc[24 + val_end : 24 + val_end + len(y_test)].values
except FileNotFoundError:
    print("Could not load original timestamp data. Creating dummy dates.")
    test_timestamps = pd.date_range("2024-01-01", periods=len(y_test), freq="H").astype(str).tolist()

def get_mape(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    non_zero = y_true != 0
    if not np.any(non_zero): return np.nan
    return np.mean(np.abs((y_true[non_zero] - y_pred[non_zero]) / y_true[non_zero])) * 100

def get_row_mapes(y_true, y_pred):
    mapes = []
    for i in range(len(y_true)):
        non_zero = y_true[i] != 0
        if np.any(non_zero):
            m = np.mean(np.abs((y_true[i][non_zero] - y_pred[i][non_zero]) / y_true[i][non_zero])) * 100
        else:
            m = np.nan
        mapes.append(m)
    return mapes

# Models Definition
seq_len = X_train.shape[1]
num_d = X_train.shape[2]
out_steps = y_train.shape[1]

def build_lstm():
    inputs = layers.Input(shape=(seq_len, num_d))
    x = layers.LSTM(64, return_sequences=False)(inputs)
    outputs = layers.Dense(out_steps)(x)
    model = models.Model(inputs=inputs, outputs=outputs, name="LSTM")
    model.compile(optimizer='adam', loss='mse')
    return model

def build_bilstm():
    inputs = layers.Input(shape=(seq_len, num_d))
    x = layers.Bidirectional(layers.LSTM(64, return_sequences=False))(inputs)
    outputs = layers.Dense(out_steps)(x)
    model = models.Model(inputs=inputs, outputs=outputs, name="BiLSTM")
    model.compile(optimizer='adam', loss='mse')
    return model

def build_attention():
    inputs = layers.Input(shape=(seq_len, num_d))
    x = layers.LayerNormalization(epsilon=1e-6)(inputs)
    attention_output = layers.MultiHeadAttention(key_dim=64, num_heads=4, dropout=0.2)(x, x)
    x = layers.Add()([attention_output, inputs])
    x2 = layers.LayerNormalization(epsilon=1e-6)(x)
    x2 = layers.Dense(64, activation="relu")(x2)
    x2 = layers.Dropout(0.2)(x2)
    x2 = layers.Dense(num_d)(x2)
    x = layers.Add()([x2, x])
    x = layers.GlobalAveragePooling1D()(x)
    x = layers.Dense(32, activation="relu")(x)
    outputs = layers.Dense(out_steps)(x)
    model = models.Model(inputs=inputs, outputs=outputs, name="MultiheadAttention")
    model.compile(optimizer='adam', loss='mse')
    return model

model_dict = {
    "LSTM": build_lstm(),
    "BiLSTM": build_bilstm(),
    "MultiheadAttention": build_attention()
}

epochs = 200
batch_size = 32

for name, model in model_dict.items():
    print(f"\n======================================")
    print(f"Training {name} Model...")
    print(f"======================================")
    
    # Train
    history = model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=epochs, batch_size=batch_size, verbose=1)
    
    # Save Loss
    train_loss = history.history['loss']
    val_loss = history.history['val_loss']
    np.save(os.path.join('LOSS', f'loss_{name}.npy'), np.array([train_loss, val_loss]))
    
    # Plot Loss
    plt.figure(figsize=(10, 6))
    plt.plot(train_loss, label='Train Loss')
    plt.plot(val_loss, label='Validation Loss')
    plt.title(f'{name} Loss over Epochs')
    plt.xlabel('Epochs')
    plt.ylabel('Loss (MSE)')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join('LOSS', f'loss_{name}.png'))
    plt.show()
    plt.close()
    
    # Save Model
    model_path = os.path.join('MODELS', f"{name}.h5")
    model.save(model_path)
    
    # Predict
    pred_train = model.predict(X_train)
    pred_test = model.predict(X_test)
    
    # MAPE
    mape_train = get_mape(y_train, pred_train)
    mape_test = get_mape(y_test, pred_test)
    
    # Save results.txt
    date_str = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    with open('RESULTS/result.txt', 'a') as f:
        f.write(f"{date_str} {name} {mape_train:.4f} {mape_test:.4f}\n")
    
    print(f"{name} MAPE Train: {mape_train:.4f} | Test: {mape_test:.4f}")
    
    # Check shape for CSV
    row_mapes = get_row_mapes(y_test, pred_test)
    
    # Convert numpy array representations to string matching lists
    lag_data_strs = [str(list(x.flatten())) for x in X_test]
    actual_strs = [str(list(y)) for y in y_test]
    pred_strs = [str(list(p)) for p in pred_test]
    
    df_res = pd.DataFrame({
        'DATE_OF_DATA_POINT': test_timestamps,
        'LAG_DATA': lag_data_strs,
        'ACTUAL_VALUE_HORIZON': actual_strs,
        'PREDICTED_VALUE_HORIZON': pred_strs,
        'MAPE_OF_THIS_DATA_POINT': row_mapes
    })
    
    df_res.to_csv(f'RESULTS/{name}.csv', index=False)
    print(f"Saved results to RESULTS/{name}.csv")
```
